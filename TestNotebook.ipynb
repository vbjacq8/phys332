### Linear Algebra Test

We'll use the following 3x3 matrix and 3x1 vector for this test:

$$\mathbf{A} = \begin{bmatrix}  -8 & 0 & -3\\  -3 & -8 & -2 \\  7 & 8 & 1 \\ \end{bmatrix},~~~~\vec{b}=\begin{bmatrix} -3\\4\\-4\end{bmatrix}$$

```python
A = np.array([[-8,  0, -3],
       [-3, -8, -2],
       [ 7,  8,  1]])

b = np.array([-3,  4, -4])
```

First, verify that the following operations yield the expected results:

- $\vec{b}\cdot\vec{b} = 41$
- $\mathbf{A}\cdot\vec{b} = \begin{bmatrix}  36 & -15 & 7\\ \end{bmatrix}$


```python
print(b@b)
```

```python
print(A@b)
```

Next, test that the compiled C/Fortran code that both Scipy and NumPy rely on under the hood (specifically the [LAPACK Library](https://en.wikipedia.org/wiki/LAPACK), originally written in Fortran 77) are working.  The following commands will solve for $\vec{x}$ in 

$$\mathbf{A}\cdot\vec{x} = \vec{b}$$

You should get $\vec{x} =  \begin{bmatrix}  0.15 & -0.70625 & 0.6\\ \end{bmatrix}$

```python
x = linalg.solve(A, b)
print(x)
```

Finally verify that $\mathbf{A}\cdot\vec{x}$ gives us back $\vec{b}$

```python
print(A@x)
```

## Matplotlib Test

This code will generate 10,000 random samples from a Gaussian distribution and plot them as a normalized histogram.  It will also plot the pdf (probabilitiy density function) on top of the histogram.

Note that this should appear *inline* in this Notebook.  If it doesn't, you'll need to root around and find another way to generate the plots (this is the most likely reason you'll have to go to the lab).

```python
rng = np.random.default_rng(42)
samples = rng.normal(size=10000)

plt.hist(samples,density=True,bins=25);
plt.xlabel("X")
plt.ylabel("PDF")

x = np.linspace(-5, 5, 201)
pdf = np.exp(-x**2 / 2) / np.sqrt(2 * np.pi)
plt.plot(x,pdf,lw=3)
```
